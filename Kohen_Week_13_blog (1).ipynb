```python
#import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.io as pio
#load dataset
df= pd.read_excel("applications.xlsx")
```

```python
#task 1: Bar Plot & compare number of appilcations by each college 
college_counts= df['College'].value_counts()

#Matplotlib
plt.figure(figsize=(8,5))
plt.bar(college_counts.index, college_counts.values, color=['blue','green','red'])
plt.title('Applications by College')
plt.xlabel('College')
plt.ylabel('Number of Applications')
plt.show()

#Seaborn 
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='College')
plt.title('Applications by College')
plt.show()

#Plotly Express
college_counts = df['College'].value_counts()

fig = px.bar(
    x=college_counts.index,
    y=college_counts.values,
    color_discrete_sequence=['pink'],
    title='Applications by College',
    labels={'x': 'College',
            'y': 'Number of Applications'}
)
fig.show()
#college with most applications
print(df["College"].value_counts())
#the college with the most applications is Arts & Letters with 6964 applicants 
```

```python

#Task 2: Histrogram SAT_ACT scores, what is the most common score range?

#Matplotlib
plt.figure(figsize=(8,5))
plt.hist(df['SAT_ACT'], bins=20, color='purple', edgecolor='black')
plt.title('Distribution of SAT_ACT Scores')
plt.xlabel('SAT_ACT Score')
plt.ylabel('Frequency')
plt.show()

#Seaborn
plt.figure(figsize=(8,5))
sns.histplot(df['SAT_ACT'], bins=20)
plt.title('Distribution of SAT_ACT Scores')
plt.show()

#Plotly
fig= px.histogram(df, x='SAT_ACT', 
                  nbins=20,
                  title='Distribution of SAT_ACT Scores')
fig.show()
#most common score range
#most common score is 1110 to 1190
```

```python
#Task 3: Scatter Plot; Is there a positive, negative, or no clear relationship?

#Matplotlib
plt.figure(figsize=(8,5))
plt.scatter(df['Extracurricular_Hours'], df['SAT_ACT'], alpha=0.5, color='teal')
#trendline
x = df['Extracurricular_Hours']
y = df['SAT_ACT']

m, b = np.polyfit(x, y, 1)
plt.plot(
    x,
    m* x + b, color='black',)
plt.title('SAT_ACT vs Extracurricular Hours')
plt.xlabel('Extracurricular Hours')
plt.ylabel('SAT_ACT')
plt.show()

#Seaborn
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df, x='Extracurricular_Hours', y='SAT_ACT')
plt.title('SAT_ACT vs Extracurricular Hours')
plt.show()

#Plotly
fig= px.scatter(
    df,
    x='Extracurricular_Hours',
    y='SAT_ACT',
    trendline='ols',
    color_discrete_sequence=['red'],
    title='SAT_ACT vs Extracurricular Hours',
    width=800,
    height=500
)
fig.show()
#correlation
#the correlation is very close to 0, so there is no clear correlation
```
